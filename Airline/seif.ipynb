```python
# Define functions to create relationships
def create_took_relationship(tx, record_locator, feedback_ID):
    tx.run("""
        MATCH (p:Passenger {record_locator: $record_locator})
        MATCH (j:Journey {feedback_ID: $feedback_ID})
        MERGE (p)-[:TOOK]->(j)
    """, record_locator=record_locator, feedback_ID=feedback_ID)

def create_on_relationship(tx, feedback_ID, flight_number):
    tx.run("""
        MATCH (j:Journey {feedback_ID: $feedback_ID})
        MATCH (f:Flight {flight_number: $flight_number})
        MERGE (j)-[:ON]->(f)
    """, feedback_ID=feedback_ID, flight_number=flight_number)

def create_departs_from_relationship(tx, flight_number, origin_station_code):
    tx.run("""
        MATCH (f:Flight {flight_number: $flight_number})
        MATCH (a:Airport {station_code: $station_code})
        MERGE (f)-[:DEPARTS_FROM]->(a)
    """, flight_number=flight_number, station_code=origin_station_code)

def create_arrives_at_relationship(tx, flight_number, destination_station_code):
    tx.run("""
        MATCH (f:Flight {flight_number: $flight_number})
        MATCH (a:Airport {station_code: $station_code})
        MERGE (f)-[:ARRIVES_AT]->(a)
    """, flight_number=flight_number, station_code=destination_station_code)
```

```python
# Create relationships from CSV data
with driver.session() as session:
    for index, row in df.iterrows():
        # (Passenger)-[:TOOK]->(Journey)
        session.execute_write(create_took_relationship, row['record_locator'], row['feedback_ID'])
        
        # (Journey)-[:ON]->(Flight)
        session.execute_write(create_on_relationship, row['feedback_ID'], row['flight_number'])
        
        # (Flight)-[:DEPARTS_FROM]->(Airport)
        session.execute_write(create_departs_from_relationship, row['flight_number'], row['origin_station_code'])
        
        # (Flight)-[:ARRIVES_AT]->(Airport)
        session.execute_write(create_arrives_at_relationship, row['flight_number'], row['destination_station_code'])
        
        if index % 100 == 0:
            # debug print
            print(f"Created relationships for {index} rows...")
```
