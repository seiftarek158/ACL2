## Strategy for handling records with same Start_Address and End_Address

Since these records have the same start and end location, we have a few options:
1. Mark them as "Invalid Route" or "Same Location"
2. Extract city/country from the address and create a route like "City-City" or "Country-Country"
3. Drop these records as they don't represent valid flight routes

````python
# Option 1: Extract country from address and create route format
# Let's see what the addresses look like
def extract_country(address):
    """Extract the country (last part) from the address"""
    if pd.isna(address):
        return None
    # Country is typically the last part after the last comma
    parts = str(address).split(',')
    return parts[-1].strip() if parts else None

# Test on the same start/end address records
asr_with_same_start_and_end_address['Country'] = asr_with_same_start_and_end_address['Start_Address'].apply(extract_country)
print(asr_with_same_start_and_end_address['Country'].value_counts())
print("\nSample:")
print(asr_with_same_start_and_end_address[['Start_Address', 'Country']].head())
````

````python
# Solution: Populate Route field for records with same start and end address
# For these cases, we'll create a route in the format "Country-Country" or mark as "Invalid Route"

# Method 1: Mark as "Same Location" or extract city information
def create_route_for_same_location(row):
    """Create a route identifier for records with same start/end"""
    if pd.isna(row['Start_Address']):
        return None
    
    # Extract city (usually second-to-last or third-to-last component)
    parts = str(row['Start_Address']).split(',')
    
    # Get country (last part)
    country = parts[-1].strip() if len(parts) > 0 else "Unknown"
    
    # Try to get city/region (work backwards to find meaningful location)
    city = parts[0].strip() if len(parts) > 0 else "Unknown"
    
    # Create route format: "City-City" to indicate same location
    return f"{city}-{city}"

# Apply to the original dataframe for rows with same start and end
mask = (asr['Start_Address'] == asr['End_Address']) & (asr['Start_Address'].notna())
asr.loc[mask, 'Route'] = asr.loc[mask].apply(create_route_for_same_location, axis=1)

# Check results
print(f"Updated {mask.sum()} records with same start and end address")
print("\nSample of updated routes:")
print(asr.loc[mask, ['Start_Address', 'End_Address', 'Route']].head(10))
````

## Using NLP to Extract Routes from Review Content

We can analyze the review text to extract origin and destination information using:
1. Pattern matching for common flight route phrases
2. Named Entity Recognition (NER) for location detection
3. Regex patterns for airport codes (e.g., LHR, JFK)

````python
# First, let's check what columns are available and examine review content
print("Dataset columns:")
print(asr.columns.tolist())
print("\n" + "="*80)

# Check for review content column (might be named 'review', 'Review_Content', 'content', etc.)
review_columns = [col for col in asr.columns if 'review' in col.lower() or 'content' in col.lower() or 'text' in col.lower()]
print(f"\nPotential review columns: {review_columns}")

# Let's look at records with same start/end address and see their review content
if review_columns:
    review_col = review_columns[0]  # Use the first review column found
    print(f"\nUsing column: '{review_col}'")
    print("\nSample reviews from records with same start/end address:")
    sample_reviews = asr_with_same_start_and_end_address[[review_col]].head(5)
    for idx, row in sample_reviews.iterrows():
        print(f"\n--- Review {idx} ---")
        print(row[review_col][:500] if pd.notna(row[review_col]) else "No review text")
        print()
else:
    print("\nNo review content column found!")
````

````python
# Let's create NLP-based route extraction functions
import re

def extract_route_patterns(text):
    """Extract route information from review text using pattern matching"""
    if pd.isna(text):
        return None, None, None
    
    text = str(text).lower()
    
    # Pattern 1: "from X to Y" or "X to Y"
    pattern1 = r'(?:from|flying from)\s+([A-Za-z\s]+?)\s+to\s+([A-Za-z\s]+?)(?:\s|\.|\,|$)'
    
    # Pattern 2: Airport codes (3 letters uppercase)
    pattern2 = r'\b([A-Z]{3})\b.*?\b([A-Z]{3})\b'
    
    # Pattern 3: "London to Paris" style
    pattern3 = r'\b([A-Z][a-z]+(?:\s[A-Z][a-z]+)?)\s+to\s+([A-Z][a-z]+(?:\s[A-Z][a-z]+)?)\b'
    
    # Try pattern matching on original case text
    original_text = str(text)
    
    # Try airport codes first (most reliable)
    match_codes = re.search(pattern2, original_text)
    if match_codes:
        return match_codes.group(1), match_codes.group(2), 'airport_code'
    
    # Try city names pattern
    match_cities = re.search(pattern3, original_text)
    if match_cities:
        return match_cities.group(1).strip(), match_cities.group(2).strip(), 'city_name'
    
    # Try "from X to Y" pattern
    match_from_to = re.search(pattern1, text)
    if match_from_to:
        return match_from_to.group(1).strip(), match_from_to.group(2).strip(), 'from_to_pattern'
    
    return None, None, None

# Test on a sample
print("Testing route extraction on sample reviews:\n")
test_sample = asr_with_same_start_and_end_address[['Review_content']].head(20)

for idx, row in test_sample.iterrows():
    origin, destination, pattern_type = extract_route_patterns(row['Review_content'])
    if origin or destination:
        print(f"Index {idx}: {origin} → {destination} (matched: {pattern_type})")
        print(f"  Review snippet: {str(row['Review_content'])[:100]}...")
        print()
````

## Advanced NLP Approach: Using spaCy for Named Entity Recognition

For even better extraction, we can use spaCy's NER to identify location entities (GPE - GeoPolitical Entity).
This can help us extract city and country names more accurately from the review text.

````python
# Enhanced extraction using spaCy with improved location detection
import spacy

# Load spaCy model
try:
    nlp = spacy.load('en_core_web_sm')
    
    def extract_route_with_spacy(text):
        if pd.isna(text):
            return None, None, 'no_text'
        
        doc = nlp(str(text)[:2000])  # Increased text length for better context
        
        # Extract all location entities (GPE = GeoPolitical Entity, LOC = Location)
        locations = [ent.text for ent in doc.ents if ent.label_ in ['GPE', 'LOC']]
        
        # Remove duplicates while preserving order
        seen = set()
        unique_locations = []
        for loc in locations:
            loc_lower = loc.lower()
            if loc_lower not in seen:
                seen.add(loc_lower)
                unique_locations.append(loc)
        
        # If we have at least 2 locations, return the first two
        if len(unique_locations) >= 2:
            return unique_locations[0], unique_locations[1], f'spacy_ner_{len(unique_locations)}_locations'
        elif len(unique_locations) == 1:
            # Only one location found - try fallback to regex patterns
            return unique_locations[0], None, 'spacy_partial'
        else:
            # No locations found
            return None, None, 'spacy_no_locations'
    
    # Apply to records that weren't extracted by regex
    not_extracted_mask = mask_same_location & asr['NLP_extracted_origin'].isna()
    
    print(f"Attempting spaCy extraction on {not_extracted_mask.sum()} remaining records...")
    print(f"This may take a while...\n")
    
    # Apply spaCy extraction
    spacy_results = asr.loc[not_extracted_mask, 'Review_content'].apply(extract_route_with_spacy)
    
    # Update the dataframe with spaCy results
    asr.loc[not_extracted_mask, 'NLP_extracted_origin'] = spacy_results.apply(lambda x: x[0])
    asr.loc[not_extracted_mask, 'NLP_extracted_destination'] = spacy_results.apply(lambda x: x[1])
    asr.loc[not_extracted_mask, 'NLP_extraction_method'] = spacy_results.apply(lambda x: x[2])
    
    # Count new successful extractions
    new_successful = not_extracted_mask & asr['NLP_extracted_origin'].notna() & asr['NLP_extracted_destination'].notna()
    
    print(f"✓ spaCy extraction complete!")
    print(f"  Previously extracted by regex: {successful_mask.sum()}")
    print(f"  Newly extracted by spaCy: {new_successful.sum()}")
    print(f"  Partial extractions (only origin): {(not_extracted_mask & asr['NLP_extracted_origin'].notna() & asr['NLP_extracted_destination'].isna()).sum()}")
    print(f"  Still no extraction: {(not_extracted_mask & asr['NLP_extracted_origin'].isna()).sum()}")
    
    # Update the successful_mask to include spaCy results
    successful_mask_updated = mask_same_location & asr['NLP_extracted_origin'].notna() & asr['NLP_extracted_destination'].notna()
    
    print(f"\n{'='*80}")
    print(f"TOTAL SUCCESSFUL EXTRACTIONS (Regex + spaCy): {successful_mask_updated.sum()}")
    print(f"Success rate: {(successful_mask_updated.sum()/mask_same_location.sum())*100:.1f}%")
    print(f"{'='*80}\n")
    
    # Show sample of spaCy extractions
    if new_successful.sum() > 0:
        print("Sample of newly extracted routes via spaCy:\n")
        sample_spacy = asr[new_successful][['NLP_extracted_origin', 'NLP_extracted_destination', 
                                             'NLP_extraction_method', 'Review_content']].head(5)
        for idx, row in sample_spacy.iterrows():
            print(f"Index {idx}:")
            print(f"  Extracted: {row['NLP_extracted_origin']} → {row['NLP_extracted_destination']}")
            print(f"  Method: {row['NLP_extraction_method']}")
            print(f"  Review: {str(row['Review_content'])[:120]}...")
            print()
    
except Exception as e:
    print(f"Error: {e}")
    print("\nMake sure spaCy is installed:")
    print("  pip install spacy")
    print("  python -m spacy download en_core_web_sm")
````

````python
# Helper function to get records by extraction status
def get_records_by_status(status='extracted'):
    """
    Get records based on NLP extraction status
    
    Parameters:
    - status: 'extracted' (successfully extracted), 
              'failed' (same start/end but couldn't extract),
              'problematic' (all records with same start/end),
              'valid' (records with different start/end)
    """
    if status == 'extracted':
        return asr[successful_mask].copy()
    elif status == 'failed':
        failed_mask = mask_same_location & asr['NLP_extracted_origin'].isna()
        return asr[failed_mask].copy()
    elif status == 'problematic':
        return asr[mask_same_location].copy()
    elif status == 'valid':
        return asr[~mask_same_location].copy()
    else:
        print("Invalid status. Use: 'extracted', 'failed', 'problematic', or 'valid'")
        return None

# Examples of how to use:
print("="*80)
print("HELPER FUNCTION: get_records_by_status()")
print("="*80)
print("\nUsage examples:")
print("  extracted_records = get_records_by_status('extracted')  # Successfully extracted")
print("  failed_records = get_records_by_status('failed')        # Failed to extract")
print("  all_problematic = get_records_by_status('problematic')  # All same start/end")
print("  valid_records = get_records_by_status('valid')          # Normal records")

print("\n" + "="*80)
print("Quick Stats:")
print("="*80)
print(f"Successfully extracted: {successful_mask.sum()} records")
print(f"Failed extraction: {(mask_same_location & asr['NLP_extracted_origin'].isna()).sum()} records")
print(f"Total problematic: {mask_same_location.sum()} records")
print(f"Valid records: {(~mask_same_location).sum()} records")
````
